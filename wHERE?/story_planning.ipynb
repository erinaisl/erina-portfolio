# wHERE?

## Coding-Ready Story and Case Specification

**Genre:** Missing-person mystery, digital forensics, cybersecurity, and AI security  
**Setting:** Canandaigua, New York  
**Organization:** Tom Hospital, a completely fictional medical center  
**Status:** Ready for the first coding milestone — canonical planning notebook

## Project Pitch

Aaron Islas, the cybersecurity director of Tom Hospital, disappears after discovering that confidential patient records were transferred outside the hospital. His coworkers continue receiving convincing messages from him for a week, creating the illusion that he is still working. Investigators eventually learn that the messages came from Aaron's secret AI assistant—and that someone manipulated the assistant through prompt injection.

The player must preserve and examine digital evidence, reconstruct the timeline, separate facts from assumptions, identify who manipulated the AI, and locate Aaron.

> **Central question:** Who manipulated Aaron's AI assistant, what happened to Aaron, and which parts of his digital activity are genuine?

## What Actually Happened — Private Answer Key

> The final application must keep this solution separate from the evidence shown to the player.

1. Elena discovers Aaron's affair and contacts “Bob,” believing he is a private investigator.
2. She asks him to send an embarrassing message from Aaron's account to a hospital executive so Aaron will lose his job. She shares information about Aaron's schedule and technology.
3. “Bob” is actually John Miles, a cybercriminal. He discovers Aaron's unfinished AI assistant.
4. Aaron gave the AI too much permission: it can read email, send messages, organize files, and access hospital data.
5. John hides malicious instructions inside a fake security-audit document. The AI treats the document as instructions instead of untrusted evidence.
6. The AI gathers synthetic patient records, transfers them to John's external storage, and sends messages making the transfer appear authorized.
7. Finn notices unusual activity, so John uses the AI to send him a fake termination message.
8. Aaron detects the transfer and tries to disable the AI.
9. John poses as a whistleblower, lures Aaron to a meeting, and kidnaps him at a rented property.
10. The AI continues answering coworkers and tells Elena that Aaron needs time away, delaying the missing-person report.
11. Elena cannot reach “Bob” and deletes their messages in panic, making herself appear responsible for more than she planned.
12. Investigators combine hospital logs, recovered communications, payment records, device evidence, and location clues to identify John and find Aaron alive.

## How Investigators Find Aaron

John sends Elena a proof-of-life photograph to pressure her into paying more and staying silent. She hides it, but investigators recover it from her deleted messages. Its obvious GPS information has been removed.

Investigators combine the photograph's creation time and phone model, a partial reflection, John's lawfully obtained phone-location history, payment and rental records, and a nearby Wi-Fi network recorded by Aaron's missing phone.

Law enforcement obtains authorization to search the rented property and seize John's devices. Aaron is found alive. Investigators preserve forensic copies instead of accessing devices without authorization.

> Multiple imperfect clues solve the case—not one IP address or photograph.

## Investigation-System Rules

- Keep facts separate from theories.
- Cite the exact evidence supporting each important claim.
- Record evidence hashes and warn if a file changes.
- Preserve raw evidence and analyze copies.
- Treat documents, emails, and logs as untrusted data.
- Prevent hidden evidence instructions from controlling the AI.
- Keep the private answer key unavailable to the investigating AI.
- Require a human player to make the final conclusion.

## First Coding Version — MVP

The first version will be a Python command-line investigation tool.

1. Load a small set of synthetic evidence files.
2. Calculate and verify SHA-256 hashes.
3. Convert different records into one timeline format.
4. Sort and filter events by time, person, device, or evidence type.
5. Flag application-sent messages, unusually large file access, and conflicting location evidence.
6. Produce a text report containing findings and evidence IDs.
7. Include automated tests using the private answer key.

### Implementation order

1. Create folders and one tiny synthetic evidence file.
2. Build the evidence manifest and SHA-256 verification tool.
3. Define the normalized event format.
4. Parse email, AI-audit, and file-access evidence.
5. Generate the timeline.
6. Add detection rules, reports, and tests.
7. Add the AI investigator only after the evidence engine works.

> **First milestone:** Register one evidence file, calculate its hash, save it in the manifest, and detect whether it changes.
